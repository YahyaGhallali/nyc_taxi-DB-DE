```python
base = "/Volumes/workspace/taxi_raw/landing"
df = spark.read.parquet(f"{base}/yellow/yellow_tripdata_2025-01.parquet")
df.printSchema()
```

```python
from pyspark.sql import functions as F

display(df.withColumn("pickup_month", F.date_format("tpep_pickup_datetime", "yyyy-MM"))
   .groupBy("pickup_month").count()
   .orderBy("pickup_month")
   .show())
```

```python
df_by_month = (df.withColumn("pickup_month", F.date_format("tpep_pickup_datetime", "yyyy-MM")).withColumn("dropoff_month", F.date_format("tpep_dropoff_datetime", "yyyy-MM")))

df_january = df_by_month.filter((F.col("pickup_month") == "2025-01") & (F.col("dropoff_month") == "2025-01"))

df_january_droped = df_january.drop("pickup_month", "dropoff_month")

display(df_january_droped)
```

```python
df.write.mode("overwrite").parquet(f"{base}/yellow/yellow_tripdata_2025-01_raw.parquet")
```
